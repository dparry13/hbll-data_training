## Analyzing the Data

During this workshop, we will learn six common pandas operations:

### Selecting columns

Suppose you are planning a strange-but-fun road trip. Select a few useful
columns: the state name, climate, UFO sightings, waffle interest, and hiking
interest.

```python
census_df[
    [
        "state",
        "avg_temp_f",
        "annual_sunny_days",
        "ufo_sightings_per_100k",
        "waffles",
        "hiking",
    ]
]
```

To see every state that has a Waffle House:

```python
census_df[census_df["has_waffle_house"]]
```

### Creating columns

The dataset includes trend scores for waffles, pancakes, and donuts. Create a
simple `Brunch_score` by adding them together. This will add a column at the end of your table.

```python
census_df = census_df.assign(
    Brunch_score=(
        census_df["waffles"]
        + census_df["pancakes"]
        + census_df["donuts"]
    )
)

census_df.sort_values("Brunch_score", ascending=False)[
    [
        "state",
        "Brunch_score",
    ]
]
```

`.assign()` returns a new DataFrame. It does not overwrite `census_df` unless
you save the result back into a variable.

### Summarizing groups

The states are grouped into personality profiles. Find the average temperature,
average annual sunshine, average UFO sightings, and number of states in each
profile. The command `.agg` is how you can apply one or more aggregate functions like `mean` and `sum` to the data.

```python
census_df.groupby("personality_profile", as_index=False).agg(
    average_temperature_f=("avg_temp_f", "mean"),
    average_sunny_days=("annual_sunny_days", "mean"),
    average_ufo_sightings=("ufo_sightings_per_100k", "mean"),
    number_of_states=("state", "size"),
)
```

Use `.groupby()` to split rows into groups before calculating a summary. For
example, find the average UFO-sighting rate for each personality profile:

```python
(
    census_df.groupby("personality_profile", as_index=False)
    .agg(average_ufo_sightings=("ufo_sightings_per_100k", "mean"))
)
```

NOTE: Keep in mind that seeing personality profile against the average UFO_sighting does not mean that you can conclude Relaxed and Creative personalities are more likely to see a UFO. There may be other factors that we have not considered, so it is important to do more tests to find other variables that could connect the data.

## More Comparison Operators

The following operators each return `True` or `False`:

- `!=` checks inequality
- `>` checks whether the left value is greater
- `<` checks whether the left value is less
- `>=` checks whether the left value is greater than or equal
- `<=` checks whether the left value is less than or equal

Notice that Colab only displays the last value.

```python
"hiking" != "skiing"
51 > 15
1000 < 1000
41 >= 41.5
50 <= 300
```

## Combining pandas Operations

You can combine filtering, selecting, creating columns, and grouping. For
example, find the average pizza and hiking trend scores for states with an NBA
team, grouped by personality profile.

**Hint:** Filter to NBA-team states first; then group by profile and calculate
two means with `.agg()`.


```python
(
    census_df[census_df["has_nba_team"]]
    .groupby("personality_profile", as_index=False)
    .agg(
        average_pizza_score=("pizza", "mean"),
        average_hiking_score=("hiking", "mean"),
        number_of_states=("state", "size"),
    )
)
```

Among states with a Waffle House, find the state with the largest waffle
trend score.

**Hint:** Filter first. `.idxmax()` finds the row label of the largest waffle
score; `.loc[...]` retrieves that row.


```python
# COMPLETED CODE BELOW
waffle_house_states = census_df[census_df["has_waffle_house"]]
waffle_house_states.loc[
    waffle_house_states["waffles"].idxmax(),
    ["state", "waffles", "pancakes", "donuts"],
]
```

For coastal states, select their state name, personality profile, annual

**Hint:** Use a coastal Boolean mask for the rows and a list of the four names

Create an `Outdoor_adventure` column that is `True` when a state has at
least 20 hiking-trend points and at least one national park. Then count outdoor
adventure states in each personality profile.

**Hint:** Put each `>=` comparison in parentheses and combine them with `&`.
Group by profile, then sum the new Boolean column to count `True` values.


```python
# COMPLETED CODE BELOW
(
    census_df.assign(
        Outdoor_adventure=(
            (census_df["hiking"] >= 20)
            & (census_df["national_parks_count"] >= 1)
        )
    )
    .groupby("personality_profile", as_index=False)
    .agg(
        outdoor_adventure_states=("Outdoor_adventure", "sum"),
        average_hiking_score=("hiking", "mean"),
    )
)
```
